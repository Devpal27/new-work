```python
import json

# Step 1: Activities column ko JSON parse karo (list of dicts banega har row mein)
df['Activities_parsed'] = df['Activities'].apply(json.loads)

# Step 2: Explode karo - har activity ke liye alag row ban jayegi
df_exploded = df.explode('Activities_parsed').reset_index(drop=True)

# Step 3: Ab har row mein ek hi dict hai, usse NIC5DigitId aur Description nikaalo
df_exploded['NIC5DigitId'] = df_exploded['Activities_parsed'].apply(lambda x: x.get('NIC5DigitId'))
df_exploded['Description'] = df_exploded['Activities_parsed'].apply(lambda x: x.get('Description'))

# Step 4: Cleanup - purani columns hata do
df_exploded.drop(columns=['Activities', 'Activities_parsed'], inplace=True)

# Check karo shape - rows badhengi kyunki multi-activity wali rows split ho gayi
print("Original shape:", df.shape)
print("Exploded shape:", df_exploded.shape)
df_exploded.head()
```
